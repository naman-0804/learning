```python
import nltk
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
```

```python
corpus="""Hello welcome , My name is naman. I am from vit chennai doing btech for last 4 years.
I have been placed in Infosys and TCS digital , but i took only TCS as it had a higher pay than infosys.
My degree is over and i am waiting for joining. Thanks."""
print(corpus)
```

```python
#corpus to document
document=sent_tokenize(corpus)
print(document)
type(document)
```

```python
#corpus to word
word_tokenize(corpus)
```

```python
#document to word
for sentence in document:
  print(word_tokenize(sentence))
```
