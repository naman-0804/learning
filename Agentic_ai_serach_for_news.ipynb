```python
#!pip install -q -U langchain langchain-google-genai langchain-community duckduckgo-search

import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
from langchain_community.tools import DuckDuckGoSearchRun

os.environ["GOOGLE_API_KEY"] = userdata.get("GEMINI_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0,
    max_output_tokens=100
)

search_tool = DuckDuckGoSearchRun()

agent = create_agent(
    model=llm,
    tools=[search_tool],
    system_prompt="""
Search current AI news and return any 1 developments.
Keep the total response under 100 words.
and tell about it in 20 words
"""
)

query = "Find the any 1 developments in AI agents and agentic AI."

result = agent.invoke({
    "messages": [{"role": "user", "content": query}]
})

print(result["messages"])
```

### Explanation of the Agentic AI News Summarizer

This notebook demonstrates a modern **Agentic Workflow** using LangChain and Google's Gemini model. Here is a breakdown of the components:

1.  **Environment Setup**: We install `langchain-google-genai` and `duckduckgo-search`. The `os.environ` line sets up the API key retrieved from Colab's secret manager (`GEMINI_API_KEY`).
2.  **LLM Configuration**: We initialize `ChatGoogleGenerativeAI` using the `gemini-1.5-flash` model. The `temperature=0` ensures the output is deterministic and factual, while `max_output_tokens=100` keeps the response concise.
3.  **Search Tool**: The `DuckDuckGoSearchRun()` object acts as a 'tool' that allows the AI to leave its static training data and browse the live internet for current events.
4.  **Agent Logic**: The `create_agent` function combines the LLM and the Search Tool. The `system_prompt` defines the rules: it instructs the agent to search for exactly one development and summarize it briefly.
5.  **Execution**: When `agent.invoke` is called, the model doesn't just guess; it analyzes the query, decides to run a search command, parses the search results, and finally formats the answer based on the prompt's constraints.
