### **PorterStemmer** Gives issues in words like history ,congratulations etc

```python
from nltk.stem import PorterStemmer
```

```python
words=["eating","eats","eaten","ate","adjustable","rafting","ability","meeting","history","goes"] #look at history
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
  print(word+" ----> "+stemming.stem(word))
```

### **RegexpStemmer class**

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|ed$|e$|able$|s$',min=4) #dollar means at last if removed then anywhere removes
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeat')
```

### **Snowball Stemmer class**

```python
from nltk.stem import SnowballStemmer
```

```python
SnowballStemmer=SnowballStemmer('english')
```

```python
for word in words:
  snowball_stemmer=SnowballStemmer.stem(word)
  print(word+" ----> "+snowball_stemmer)
```

```python
stemming.stem('fairly'),stemming.stem('sportingly')
```

```python
SnowballStemmer.stem('fairly'),SnowballStemmer.stem('sportingly') # so it worked here better than porterstemmer but still has issues
```
