```python

import pandas as pd
messages=pd.read_csv('SMSSpamCollection.txt', sep='\t', names=["label", "message"])
```

```python
#!pip install nltk
#nltk.download('punkt_tab')
import nltk
#nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
leammatizer=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(len(messages)):
    review=re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[leammatizer.lemmatize(word) for word in review ]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words=[]
for sent in corpus:
    words.append(simple_preprocess(sent))
```

```python
words
```

```python
##LEts train word2vec model on our own data
model=gensim.models.Word2Vec(words,epochs=100 ,vector_size=100)
model.wv.index_to_key #all the words in the vocabulary
```

```python
model.epochs
```

```python
model.wv.similar_by_word('good')
```

```python
def avg_word2vec(doc):
    vectors = [model.wv[word] for word in doc if word in model.wv.index_to_key]
    if not vectors:
        return np.zeros(model.vector_size, dtype='float32')
    return np.mean(vectors, axis=0).astype('float32')
```
