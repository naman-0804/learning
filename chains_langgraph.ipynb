```python
from dotenv import load_dotenv
load_dotenv()
import os
```

```python
from langchain_core.messages import AIMessage,HumanMessage
from pprint import pprint

messages=[AIMessage(content=f"Please tell me how can I help",name="LLM Model")]
messages.append(HumanMessage(content=f"I want to learn coding",name="Naman"))
messages.append(AIMessage(content=f"Which programming language?",name="LLM Model"))
for message in messages:
  message.pretty_print()
```

```python
import google.generativeai as genai
from google.colab import userdata

# Securely retrieve the API key from Colab secrets
GOOGLE_API_KEY = userdata.get('GEMINI_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY)

# Initialize the model
model = genai.GenerativeModel('gemini-3.5-flash')
print("Gemini model initialized successfully.")
```

```python
# Convert LangChain messages to Gemini content format
gemini_history = []
for msg in messages:
    role = "model" if isinstance(msg, AIMessage) else "user"
    gemini_history.append({"role": role, "parts": [msg.content]})

# The last message is from the 'model' in your list, so we'll start a chat
# and ask for a response to the current state if needed,
# or simply generate based on the sequence.

# To get a NEW response from the model, we ensure the last role was 'user'.
# In your current list, the last message is an AIMessage ("Which programming language?").
# I will prompt the model to continue or provide a completion.

chat = model.start_chat(history=gemini_history[:-1])
response = chat.send_message(gemini_history[-1]['parts'][0])

print(f"AI Response: {response.text}")
```
