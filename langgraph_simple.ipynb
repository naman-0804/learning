```python
#BUILD A SIMPLE WORKFLOW OR GRAPH USING LANG GRAPH
```

```python
from typing_extensions import TypedDict
class State(TypedDict):
  graph_info: str
```

```python
def start_play(state:State): #Entry
  print('Start_play node has been called')
  return{"graph_info":state['graph_info']+"I am planning to play"}
def cricket(state:State):
  print('Cricket node has been called')
  return{"graph_info":state['graph_info']+"cricket"}
def football(state:State):
  print('Football node has been called')
  return{"graph_info":state['graph_info']+"football"}
```

```python
import random
from typing import Literal

def random_play(state:State)->Literal['cricket','football']:
  graph_info=state=['graph_info']
  if random.random()>=0.5:
    return "cricket"
  else:
    return "football"
```

```python
from re import I
from IPython.display import Image,display
from langgraph.graph import StateGraph,START,END

#Build graph
graph=StateGraph(State)
#ADD nodes
graph.add_node("start_play",start_play)
graph.add_node("cricket",cricket)
graph.add_node("football",football)
#Schedule the flow of the graph
graph.add_edge(START,"start_play")
graph.add_conditional_edges("start_play",random_play)
graph.add_edge("cricket",END)
graph.add_edge("football",END)
#cOMPILE THE GRAPH
graph_builder=graph.compile()
#view
display(Image(graph_builder.get_graph().draw_mermaid_png()))
```

```python
#Graph invocation
graph_builder.invoke({"graph_info":"Hey My name is naman "})
```
