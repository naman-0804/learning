```python
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader('attention(1).pdf')
docs=loader.load()
docs
```

```python
#How to recursively split text by character
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_spiltter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_docs=text_spiltter.split_documents(docs)
print(final_docs[0])
print('/n')
print(final_docs[1])
#see overallped chunking here
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=5)
speech=""
with open ("speech.txt") as f:
  speech=f.read()
text=text_splitter.create_documents([speech])
print(text[1])
```
